```python
# ============================================
# E-COMMERCE DATASET - PREPROCESSING
# ============================================

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer

# --------------------------------------------
# 1. Load Dataset
# --------------------------------------------
file_path = "/kaggle/input/datasets/carrie1/ecommerce-data/data.csv"

df = pd.read_csv(
    file_path,
    encoding="ISO-8859-1"
)

print("Original Shape:", df.shape)

# --------------------------------------------
# 2. Remove Duplicate Rows
# --------------------------------------------
df = df.drop_duplicates()

print("After removing duplicates:", df.shape)

# --------------------------------------------
# 3. Convert InvoiceDate to Datetime
# --------------------------------------------
df["InvoiceDate"] = pd.to_datetime(
    df["InvoiceDate"],
    errors="coerce"
)

# Remove rows where date is invalid
df = df.dropna(subset=["InvoiceDate"])

# --------------------------------------------
# 4. Handle Missing Values
# --------------------------------------------

# Description missing values
df["Description"] = df["Description"].fillna("Unknown")

# CustomerID missing values
df["CustomerID"] = df["CustomerID"].fillna(-1)

# --------------------------------------------
# 5. Remove Invalid Quantity
# --------------------------------------------

# Negative quantity usually represents returns/cancellations.
# For a normal sales prediction project, keep only positive sales.
df = df[df["Quantity"] > 0]

# --------------------------------------------
# 6. Remove Invalid UnitPrice
# --------------------------------------------

df = df[df["UnitPrice"] > 0]

# --------------------------------------------
# 7. Create Sales Target
# --------------------------------------------

df["Sales"] = df["Quantity"] * df["UnitPrice"]

# --------------------------------------------
# 8. Extract Date Features
# --------------------------------------------

df["Year"] = df["InvoiceDate"].dt.year
df["Month"] = df["InvoiceDate"].dt.month
df["Day"] = df["InvoiceDate"].dt.day
df["DayOfWeek"] = df["InvoiceDate"].dt.dayofweek
df["Hour"] = df["InvoiceDate"].dt.hour

# --------------------------------------------
# 9. Create Weekend Feature
# --------------------------------------------

df["IsWeekend"] = (
    df["DayOfWeek"] >= 5
).astype(int)

# --------------------------------------------
# 10. Remove Original Date Column
# --------------------------------------------

df = df.drop(columns=["InvoiceDate"])

# --------------------------------------------
# 11. Remove ID Columns
# --------------------------------------------

# IDs are not useful as direct numerical features.
df = df.drop(
    columns=["InvoiceNo", "StockCode", "CustomerID"],
    errors="ignore"
)

# --------------------------------------------
# 12. Separate Features and Target
# --------------------------------------------

X = df.drop(columns=["Sales"])
y = df["Sales"]

print("\nFeatures:")
print(X.columns.tolist())

print("\nTarget: Sales")

# --------------------------------------------
# 13. Identify Categorical Columns
# --------------------------------------------

categorical_cols = X.select_dtypes(
    include=["object"]
).columns.tolist()

numeric_cols = X.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

print("\nCategorical Columns:")
print(categorical_cols)

print("\nNumerical Columns:")
print(numeric_cols)

# --------------------------------------------
# 14. One-Hot Encoding
# --------------------------------------------

preprocessor = ColumnTransformer(
    transformers=[
        (
            "cat",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            ),
            categorical_cols
        )
    ],
    remainder="passthrough"
)

# --------------------------------------------
# 15. Train-Test Split
# --------------------------------------------

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("\nTraining data:", X_train.shape)
print("Testing data :", X_test.shape)

# --------------------------------------------
# 16. Fit Preprocessor ONLY on Training Data
# --------------------------------------------

X_train_processed = preprocessor.fit_transform(X_train)

X_test_processed = preprocessor.transform(X_test)

print("\nProcessed Training Shape:")
print(X_train_processed.shape)

print("\nProcessed Testing Shape:")
print(X_test_processed.shape)

# --------------------------------------------
# 17. Final Check
# --------------------------------------------

print("\n========== PREPROCESSING COMPLETED ==========")

print("X_train:", X_train_processed.shape)
print("X_test :", X_test_processed.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)

print("\nMissing values after cleaning:")
print(df.isnull().sum().sum())

print("\nSample of cleaned data:")
display(df.head())
```

```python
# ============================================================
# E-COMMERCE SALES PREDICTION
# DECISION TREE REGRESSION - COMPLETE ONE CELL
# ============================================================

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score


# ============================================================
# 1. LOAD DATASET
# ============================================================

file_path = "/kaggle/input/datasets/carrie1/ecommerce-data/data.csv"

df = pd.read_csv(
    file_path,
    encoding="ISO-8859-1"
)

print("Original Dataset Shape:", df.shape)


# ============================================================
# 2. REMOVE DUPLICATES
# ============================================================

df = df.drop_duplicates()

print("After Removing Duplicates:", df.shape)


# ============================================================
# 3. CONVERT DATE
# ============================================================

df["InvoiceDate"] = pd.to_datetime(
    df["InvoiceDate"],
    errors="coerce"
)

# Remove invalid dates
df = df.dropna(subset=["InvoiceDate"])


# ============================================================
# 4. HANDLE MISSING VALUES
# ============================================================

df["Description"] = df["Description"].fillna("Unknown")

df["CustomerID"] = df["CustomerID"].fillna(-1)


# ============================================================
# 5. REMOVE INVALID SALES
# ============================================================

# Negative quantity = returns/cancelled transactions
df = df[df["Quantity"] > 0]

# UnitPrice should be greater than zero
df = df[df["UnitPrice"] > 0]


# ============================================================
# 6. CREATE SALES TARGET
# ============================================================

df["Sales"] = df["Quantity"] * df["UnitPrice"]


# ============================================================
# 7. CREATE DATE FEATURES
# ============================================================

df["Year"] = df["InvoiceDate"].dt.year

df["Month"] = df["InvoiceDate"].dt.month

df["Day"] = df["InvoiceDate"].dt.day

df["DayOfWeek"] = df["InvoiceDate"].dt.dayofweek

df["Hour"] = df["InvoiceDate"].dt.hour

df["IsWeekend"] = (
    df["DayOfWeek"] >= 5
).astype(int)


# ============================================================
# 8. DROP ORIGINAL DATE
# ============================================================

df = df.drop(
    columns=["InvoiceDate"]
)


# ============================================================
# 9. DROP ID COLUMNS
# ============================================================

df = df.drop(
    columns=[
        "InvoiceNo",
        "StockCode",
        "CustomerID"
    ],
    errors="ignore"
)


# ============================================================
# 10. FEATURES AND TARGET
# ============================================================

X = df.drop(
    columns=["Sales"]
)

y = df["Sales"]


print("\nFeatures:")
print(X.columns.tolist())

print("\nTarget: Sales")


# ============================================================
# 11. IDENTIFY CATEGORICAL COLUMNS
# ============================================================

categorical_cols = X.select_dtypes(
    include=["object"]
).columns.tolist()

numeric_cols = X.select_dtypes(
    include=np.number
).columns.tolist()


print("\nCategorical Columns:")
print(categorical_cols)

print("\nNumerical Columns:")
print(numeric_cols)


# ============================================================
# 12. TRAIN TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)


print("\nTraining Shape:", X_train.shape)
print("Testing Shape :", X_test.shape)


# ============================================================
# 13. ONE-HOT ENCODING
# ============================================================

preprocessor = ColumnTransformer(
    transformers=[
        (
            "categorical",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            ),
            categorical_cols
        )
    ],
    remainder="passthrough"
)


# ============================================================
# 14. TRANSFORM DATA
# ============================================================

X_train_processed = preprocessor.fit_transform(
    X_train
)

X_test_processed = preprocessor.transform(
    X_test
)


print("\nProcessed Training Shape:",
      X_train_processed.shape)

print("Processed Testing Shape:",
      X_test_processed.shape)


# ============================================================
# 15. DECISION TREE REGRESSION MODEL
# ============================================================

model = DecisionTreeRegressor(
    max_depth=15,
    min_samples_split=10,
    min_samples_leaf=5,
    random_state=42
)


# ============================================================
# 16. TRAIN MODEL
# ============================================================

print("\nTraining Decision Tree Regression...")

model.fit(
    X_train_processed,
    y_train
)

print("✅ Model Training Completed!")


# ============================================================
# 17. PREDICTION
# ============================================================

y_pred = model.predict(
    X_test_processed
)


# ============================================================
# 18. EVALUATION
# ============================================================

mae = mean_absolute_error(
    y_test,
    y_pred
)

mse = mean_squared_error(
    y_test,
    y_pred
)

rmse = np.sqrt(mse)

r2 = r2_score(
    y_test,
    y_pred
)


# ============================================================
# 19. RESULTS
# ============================================================

print("\n====================================")
print("DECISION TREE REGRESSION RESULTS")
print("====================================")

print(f"MAE      : {mae:.2f}")

print(f"MSE      : {mse:.2f}")

print(f"RMSE     : {rmse:.2f}")

print(f"R² Score : {r2:.4f}")


# ============================================================
# 20. ACTUAL VS PREDICTED
# ============================================================

results = pd.DataFrame({
    "Actual Sales": y_test.values[:10],
    "Predicted Sales": y_pred[:10]
})

print("\nFirst 10 Predictions:")

display(results)
```
